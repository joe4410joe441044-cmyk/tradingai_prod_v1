```python
def calc_qty(balance, risk_pct, price, sl_pct, leverage):
    risk_amount = balance * (risk_pct / 100)
    qty = risk_amount / (price * (sl_pct / 100))
    position_value = qty * price
    required_margin = position_value / leverage
    return qty, position_value, required_margin
```

```python
import ipywidgets as w
from IPython.display import display

balance = w.FloatSlider(value=100, min=10, max=1000, step=10, description="資金")
risk    = w.FloatSlider(value=5, min=1, max=10, step=1, description="リスク%")
price   = w.FloatSlider(value=80000, min=10000, max=100000, step=1000, description="価格")
sl      = w.FloatSlider(value=2, min=0.5, max=10, step=0.5, description="SL%")
lev     = w.IntSlider(value=15, min=1, max=50, step=1, description="レバ")

out = w.Output()

def update(*_):
    out.clear_output()
    qty, pos, margin = calc_qty(
        balance.value, risk.value, price.value, sl.value, lev.value
    )
    with out:
        print(f"qty: {qty:.6f} BTC")
        print(f"ポジション: {pos:,.0f} USDT")
        print(f"証拠金: {margin:,.0f} USDT")

for x in [balance, risk, price, sl, lev]:
    x.observe(update, 'value')

display(balance, risk, price, sl, lev, out)
update()
```
